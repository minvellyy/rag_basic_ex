## 코드로 index 삭제

```python
# 삭제할 embedding vector DB index 
index_name = "embedding-test"

# Pinecone에 있는 모든 인덱스를 순회하면 찾음 inex_name과 일치하는 인덱스를 찾아서 삭제

for idx in pc.list_indexes():
    # 인덱스 이름이 "quickstart"와 일치하는 경우 해당 인덱스를 삭제함
    if idx.name == index_name:
        pc.delete_index(idx.name)
```
